```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, accuracy_score


df = pd.read_csv('disaster3.csv', encoding='latin1')


print("Dataset loaded successfully!")
print(df.head())
print("-" * 30)

# 2. Data Pre-processing
# Encode categorical variables (Geo_Location, Time_Pattern, Risk_Level)
label_encoders = {}
categorical_cols = ['Geo_Location', 'Time_Pattern', 'Risk_Level']

for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le

# Define Features (X) and Target (y)
X = df.drop('Risk_Level', axis=1)
y = df['Risk_Level']

# 3. Split Dataset (Given small dataset, setting test_size small)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 4. Model Training (Random Forest)
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# 5. Prediction and Evaluation
y_pred = rf_model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n")
# Inverse transform to show original labels in report
target_names = label_encoders['Risk_Level'].classes_
print(classification_report(y_test, y_pred, target_names=target_names))

# Example: Predict risk for a new scenario
# Data: Hist=10, Rain=300, Seismic=4.0, Wind=80, Location=Coastal(0), Time=Winter(3)
new_data = [[10, 300, 4.0, 80, 0, 3]]
prediction = rf_model.predict(new_data)
predicted_risk = label_encoders['Risk_Level'].inverse_transform(prediction)
print(f"Predicted Risk for new data: {predicted_risk[0]}")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import AdaBoostClassifier
from sklearn.metrics import classification_report, accuracy_score

# 1. Load the Dataset
# Fixing UnicodeDecodeError by using 'latin1' or 'utf-8-sig'
df = pd.read_csv('disaster.csv', encoding='latin1')

# 2. Data Pre-processing
# Encode Categorical Variables
le_loc = LabelEncoder()
le_time = LabelEncoder()
le_risk = LabelEncoder()

df['geographic_location'] = le_loc.fit_transform(df['geographic_location'])
df['time_pattern'] = le_time.fit_transform(df['time_pattern'])
df['risk_level'] = le_risk.fit_transform(df['risk_level']) # Low=0, Med=1, High=2

# Define Features and Target
X = df.drop('risk_level', axis=1)
y = df['risk_level']

# Split Data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Normalization
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# 3. Model Training - Ada Boosting
# Using DecisionTreeClassifier as base estimator implicitly
model = AdaBoostClassifier(n_estimators=50, learning_rate=1.0, random_state=42)
model.fit(X_train, y_train)

# 4. Prediction and Evaluation
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred, target_names=le_risk.classes_))
```
